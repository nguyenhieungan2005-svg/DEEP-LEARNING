## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)



print("C =", C)
print("D =", D)
```

## Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:
* Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
* Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
* Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
* Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import torch
import numpy as np

# Yêu cầu 1: Tạo Tensor X 24 phần tử và reshape
X = torch.arange(24)
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)
print("X reshaped to 4x6:")
print(X_4x6)    
print("X reshaped to 2x12:")
print(X_2x12)
print("X reshaped to 2x3x4:")
print(X_2x3x4)

# Yêu cầu 2: Thao tác trên hai Tensor A và B kích thước 3x4
A = torch.randn(3, 4)
B = torch.randn(3, 4)
add_res = A + B
mul_res = A * B
div_res = A / B
mean_col = torch.mean(A, dim=0)
std_col = torch.std(A, dim=0)
print("Addition result:")
print(add_res)
print("Multiplication result:")
print(mul_res)
print("Division result:")
print(div_res)
print("Mean of each column:")
print(mean_col)
print("Standard deviation of each column:")
print(std_col)

# Yêu cầu 3: Tính đạo hàm của y = 2*x^2 + 5*x + 3 tại x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 2 * (x ** 2) + 5 * x + 3
y.backward()
# Đạo hàm thủ công: y' = 4x + 5. Tại x = 2: y' = 4*2 + 5 = 13.
# x.grad trả về giá trị 13.0, hoàn toàn trùng khớp với tính toán thủ công.

print("Value of y at x=2:", y.item())
print("Derivative of y at x=2:", x.grad.item())



# Yêu cầu 4: Huấn luyện tìm hệ số của đa thức y = 2*x^2 - 3*x + 5
poly = np.poly1d([2, -3, 5])
N = 20
X_np = np.random.rand(N, 1) * 5
Y_np = poly(X_np)

XX_np = np.hstack([X_np ** 2, X_np, np.ones_like(X_np)])
w = torch.randn(3, 1, requires_grad=True)
x_train = torch.tensor(XX_np, dtype=torch.float32)
y_train = torch.tensor(Y_np, dtype=torch.float32)

opt = torch.optim.NAdam([w], lr=0.01)
for epoch in range(2000):
    opt.zero_grad()
    y_pred = x_train @ w
    loss = torch.mean(torch.square(y_train - y_pred))
    loss.backward()
    opt.step()
print("Learned coefficients:")
print(w)

# Kết quả vector w hội tụ về xấp xỉ [2.0, -3.0, 5.0]
```
